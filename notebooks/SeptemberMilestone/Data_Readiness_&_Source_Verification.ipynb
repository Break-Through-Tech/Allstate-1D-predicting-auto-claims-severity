# Schema and Integrity



```python
#Exploring the column headers

print("Expected column headers of the dataset:" +
      "\nid,cat1 ... cat116, cont1 ... cont14, loss\n")

print("Actual column headers of the dataset:")

for i in range(len(df.columns)):
  col = df.columns[i]
  if(i % 10 == 0):
    print(col + ", \n", end='')
  else:
    print(col + ", ", end='')
```

```python
#Finding which records, if any, have missing values for the 'id' field

print(f"Number of missing 'id' values: {df['id'].isnull().sum()}\n")

missing_id_records = df[df['id'].isnull()]
if(not missing_id_records.empty):
  print("Records with missing 'id' values are:")
  display(missing_id_records)
else:
  print("No records found with missing 'id' values!")
```

```python
#Find and count duplicate rows, if any

duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}\n")

if duplicate_count != 0:
  print("Duplicate rows found:")
  display(df[df.duplicated()])
else:
  print("No duplicate rows were found.")
```

```python
# Confirm the data types of the 'cont*' fields are numeric

cont_fields = [col for col in df.columns if col.startswith('cont')]

print(df[cont_fields].dtypes)
```

```python
#Find records with numerical values not between 0 or 1

invalid_cont_fields = []

for cont_field in cont_fields:
  if((df[cont_field] < 0) | (df[cont_field] > 1)).any():
    invalid_cont_fields.append(cont_field)

if invalid_cont_fields:
  print("cont* fields with invalid values are:\n " + invalid_cont_fields)
else:
    print("There are no cont* fields with invalid values")
```

```python
#Check that the 'loss' field is appropriately structured: only numeric values, finite, and strictly positive

loss_values = df['loss']
invalid_loss_records = []

print(df['loss'].dtype)

for loss_value in loss_values:
  if np.isfinite(loss_value) == False | (loss_value < 0): #checks if a record has an inifinte value or is negative
    invalid_loss_records.append(df[loss_value])

if invalid_loss_records:
  print("Records with invalid loss values are:\n ")
  print(invalid_loss_records)
else:
    print("There are no records with invalid loss values")
```

```python
# Confirm that all cat* fields are stored as non-numeric object types

cat_fields = [col for col in df.columns if col.startswith('cat')]

print(df[cat_fields].dtypes)
```
